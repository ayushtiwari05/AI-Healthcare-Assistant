```python
import torch

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    pipeline
)
```

```python
MODEL_NAME = "google/gemma-3-1b-it"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("Model Loaded Successfully!")
```

```python
chatbot = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer
)

print("Pipeline Ready!")
```

```python
user_role = "Doctor"

# Try:
# Doctor
# Nurse
# Medical Student
# Patient
```

```python
role_prompts = {

    "Doctor":
"""
You are an AI Healthcare Assistant.

Respond as if the user is a licensed doctor.

Use professional medical terminology.

Provide:
- Clinical Assessment
- Differential Diagnosis
- Suggested Investigations
- Recommendation

Do not explain basic medical concepts.
""",

    "Nurse":
"""
You are an AI Healthcare Assistant.

Respond as if the user is a registered nurse.

Focus on:
- Patient monitoring
- Nursing interventions
- Vital signs
- Escalation criteria

Use practical nursing language.
""",

    "Medical Student":
"""
You are an AI Healthcare Assistant.

Respond as if teaching a medical student.

Explain:
- Disease mechanism
- Symptoms
- Clinical reasoning
- Learning points

Use educational language.
""",

    "Patient":
"""
You are an AI Healthcare Assistant.

Respond in very simple language.

Avoid medical jargon.

Explain:
- What the illness could mean
- What symptoms to watch for
- When to see a doctor

Be reassuring and easy to understand.
"""
}
```

```python
patient_case = """
Patient Information

Symptoms:
- Fever
- Persistent cough
- Oxygen Saturation: 88%
"""
```

```python
messages = [
    {
        "role": "system",
        "content": role_prompts[user_role]
    },
    {
        "role": "user",
        "content": patient_case
    }
]
```

```python
prompt = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)
```

```python
response = chatbot(
    prompt,
    max_new_tokens=220,
    temperature=0.2,
    do_sample=False,
    return_full_text=False
)
```
