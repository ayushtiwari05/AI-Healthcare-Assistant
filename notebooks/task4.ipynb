```python
import torch

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    pipeline
)
```

```python
MODEL_NAME = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

print("Selected Model :", MODEL_NAME)
```

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print("Tokenizer Loaded Successfully")
```

```python
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    torch_dtype=torch.float16,
    device_map="auto"
)

print("Model Loaded Successfully")
```

```python
chatbot = pipeline(
    task="text-generation",
    model=model,
    tokenizer=tokenizer
)

print("Pipeline Ready")
```

```python
few_shot_prompt = """
You are an AI Healthcare Assistant.

Your task is to convert clinical notes into structured summaries.

Example 1

Input:
Patient has cough and fever.

Output:
Symptoms:
- Cough
- Fever

Severity:
Moderate

----------------------------------------

Example 2

Input:
Patient reports severe chest pain and difficulty breathing.

Output:
Symptoms:
- Chest pain
- Difficulty breathing

Severity:
High

----------------------------------------

Example 3

Input:
Patient has headache and mild dizziness.

Output:
Symptoms:
- Headache
- Dizziness

Severity:
Mild

----------------------------------------

Now convert the following clinical note.

Input:
Patient has sore throat, fever, body pain and fatigue.

Output:
"""
```

```python
response = chatbot(
    few_shot_prompt,
    temperature=0.2,
    do_sample=True
)
```

```python
generated = response[0]["generated_text"]

answer = generated.replace(few_shot_prompt, "").strip()

print("=" * 60)
print("Structured Summary")
print("=" * 60)

print(answer)
```

```python
with open("task3_output.md", "w", encoding="utf-8") as file:
    file.write(answer)

print("Task 3 Output Saved Successfully!")
```
